```python
# ex1

from collections import Counter
import matplotlib.pyplot as plt

from collections import Counter

s_init = "bacGBABCBAGBCBAGCbacbacbacbabcaccccccccccccccccccc"

s = s_init.lower() 

counts = {char:s.count(char) for char in s}

print(counts)
```

```python

# ex2
combinations_2 = [s[i:i+2] for i in range(len(s) - 1)]
combinations_3 = [s[i:i+3] for i in range(len(s) - 2)]

print("2-combinations:", combinations_2)
print("3-combinations:", combinations_3)

freq_2 = Counter(combinations_2)
freq_3 = Counter(combinations_3)

labels_2, counts_2 = zip(*freq_2.items()) if freq_2 else ([], [])
labels_3, counts_3 = zip(*freq_3.items()) if freq_3 else ([], [])

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
    
# Left subplot: 2-Letter Combinations
ax1.bar(labels_2, counts_2, color='skyblue', edgecolor='black')
ax1.set_ylabel("Frequency")
ax1.grid(axis='y', linestyle='--', alpha=0.7)

# Right subplot: 3-Letter Combinations
ax2.bar(labels_3, counts_3, color='salmon', edgecolor='black')
ax2.set_ylabel("Frequency")
ax2.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.savefig("combinations_frequency.png")
```
