Make a software implementation in which the native design of the global sequence alignment is used to align two sequences.

s1 = 'ATCCGTTAC'
s2 = 'CGAT'

$$
 a[i,j] = MAX \begin{cases}
a[i-1, j-1] + s(xi, xj) \\
a[i-1, j] + g \\
a[i, j-1] +g
\end{cases}
$$

$$
s = \begin{cases}
+2, \text {if } x_i = x_j \\
-1, \text {if } x_i \neq x_j
$$

M = + 2 (if we have a match)
MM = -1 (if we have a mismatch)
g = -2 (gap)



```python
s1 = 'ATCCGTTAC'
s2 = 'TCCA'

M, MM, g = 2, -1, -2


def needleman_wunsch(s1, s2, M=2, MM=-1, g=-2):
    n, m = len(s1), len(s2)
    # score matrix + traceback matrix
    a = [[0] * (m + 1) for _ in range(n + 1)]
    for i in range(1, n + 1):
        a[i][0] = i * g
    for j in range(1, m + 1):
        a[0][j] = j * g

    for i in range(1, n + 1):
        for j in range(1, m + 1):
            s = M if s1[i - 1] == s2[j - 1] else MM
            a[i][j] = max(a[i - 1][j - 1] + s,
                          a[i - 1][j] + g,
                          a[i][j - 1] + g)

    # traceback
    al1, al2 = '', ''
    i, j = n, m
    while i > 0 or j > 0:
        if i > 0 and j > 0 and a[i][j] == a[i - 1][j - 1] + (M if s1[i - 1] == s2[j - 1] else MM):
            al1, al2 = s1[i - 1] + al1, s2[j - 1] + al2
            i, j = i - 1, j - 1
        elif i > 0 and a[i][j] == a[i - 1][j] + g:
            al1, al2 = s1[i - 1] + al1, '-' + al2
            i -= 1
        else:
            al1, al2 = '-' + al1, s2[j - 1] + al2
            j -= 1
    return a, al1, al2


a, al1, al2 = needleman_wunsch(s1, s2, M, MM, g)

# print the matrix
print('      -  ' + '  '.join(f'{c:>2}' for c in s2))
for i, row in enumerate(a):
    label = '-' if i == 0 else s1[i - 1]
    print(f'{label} ' + ' '.join(f'{v:>3}' for v in row))

print()
print(al1)
print(''.join('|' if x == y else ' ' for x, y in zip(al1, al2)))
print(al2)
print('Score:', a[-1][-1])
```
